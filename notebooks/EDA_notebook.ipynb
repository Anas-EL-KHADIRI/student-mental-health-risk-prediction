```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11



df=pd.read_csv("../data/raw/student_mental_health.csv")
print("importing libraries and reading the data")
```

```python
print(df.columns)
print("\n")
print(df.dtypes)
print("\n")
print(df.shape)
df.head()
```

```python
numeric_cols=["Sleep_Hours","GPA","Social_Media_Hours","Financial_Stress","Family_Support","Attendance_Rate"]
categorical_cols=["Gender","Mental_Health_Risk","Physical_Activity","Part_Time_Job","Diet_Quality","Previous_Mental_Health_History"]
```

```python
display(df.isna().sum())
print(f"nubers of duplicated is : {df.duplicated().sum()} dup")
df.drop_duplicates(inplace=True)
display(df.describe().T)
# df[numeric_cols]=df[numeric_cols].fillna(df[numeric_cols].mean(), inplace=True)
df.fillna({numeric_cols[0]: df[numeric_cols[0]].mean()}, inplace=True)
df["Diet_Quality"]=df["Diet_Quality"].fillna(df["Diet_Quality"].mode()[0])

display(df.isna().sum())
display(df.describe().T)
```

```python
df.drop(columns=["Student_ID"], inplace=True)
```

```python
risk_order = ['Low', 'Medium', 'High']
risk_colors = ['#2ecc71', '#f39c12', '#e74c3c']

counts = df['Mental_Health_Risk'].value_counts()[risk_order]
plt.pie(counts, labels=risk_order, autopct='%1.1f%%', colors=risk_colors, startangle=90, explode=(0.02, 0.02, 0.05))
plt.title("Class Balance (%)", fontweight='bold')

plt.show()
```

```python
continuous_cols = [
    'Sleep_Hours', 'Study_Hours_Per_Week', 'GPA', 'Social_Media_Hours', 
    'Screen_Time_Hours', 'Financial_Stress', 'Family_Support', 'Attendance_Rate'
]

fig, axes = plt.subplots(4, 2, figsize=(14, 14))
axes = axes.flatten()

for i, col in enumerate(continuous_cols):
    sns.histplot(df[col], kde=True, ax=axes[i], color='#3498db', bins=25)
    axes[i].set_title(f"Distribution of {col}", fontweight='bold')
    axes[i].set_xlabel(col)

plt.tight_layout()
plt.show()
```

```python

fig, axes = plt.subplots(2, 4, figsize=(16, 7))
axes = axes.flatten()

for i, col in enumerate(continuous_cols):
    sns.boxplot(y=df[col], ax=axes[i], color='#9b59b6')
    axes[i].set_title(col, fontweight='bold')

plt.tight_layout()
plt.show()
print("Outlier detection using IQR method:")
for col in continuous_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    n_outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    print(f"{col:22} : {n_outliers:2} outliers detected [Bounds: {lower_bound:.2f} to {upper_bound:.2f}]")
```

```python
risk_order = ['Low', 'Medium', 'High']
risk_colors = ["#04049E", "#860000", "#7f0c8e"]

# 1. Numerical Factors vs Mental Health Risk
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.boxplot(data=df, x='Mental_Health_Risk', y='Sleep_Hours', order=risk_order, palette=risk_colors, ax=axes[0, 0])
axes[0, 0].set_title("Sleep Hours by Risk Level", fontweight='bold')

sns.boxplot(data=df, x='Mental_Health_Risk', y='Financial_Stress', order=risk_order, palette=risk_colors, ax=axes[0, 1])
axes[0, 1].set_title("Financial Stress (1-10) by Risk Level", fontweight='bold')

sns.boxplot(data=df, x='Mental_Health_Risk', y='Family_Support', order=risk_order, palette=risk_colors, ax=axes[1, 0])
axes[1, 0].set_title("Family Support (1-10) by Risk Level", fontweight='bold')

sns.boxplot(data=df, x='Mental_Health_Risk', y='Screen_Time_Hours', order=risk_order, palette=risk_colors, ax=axes[1, 1])
axes[1, 1].set_title("Daily Screen Time (Hours) by Risk Level", fontweight='bold')

plt.tight_layout()
plt.show()

# 2. Categorical Factors vs Mental Health Risk
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(data=df, x='Previous_Mental_Health_History', hue='Mental_Health_Risk', hue_order=risk_order, palette=risk_colors, ax=axes[0])
axes[0].set_title("Previous Mental Health History vs Risk", fontweight='bold')

sns.countplot(data=df, x='Diet_Quality', order=['Poor', 'Average', 'Good'], hue='Mental_Health_Risk', hue_order=risk_order, palette=risk_colors, ax=axes[1])
axes[1].set_title("Diet Quality vs Risk", fontweight='bold')

sns.countplot(data=df, x='Physical_Activity', order=['Low', 'Moderate', 'High'], hue='Mental_Health_Risk', hue_order=risk_order, palette=risk_colors, ax=axes[2])
axes[2].set_title("Physical Activity vs Risk", fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
corr_matrix = df[numeric_cols].corr(method="spearman")

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix")
plt.show()
```

```python
from sklearn.preprocessing import OrdinalEncoder


encoder = OrdinalEncoder(
    categories=[['Low', 'Medium', 'High']]
)

df[['Mental_Health_Risk']] = encoder.fit_transform(
    df[['Mental_Health_Risk']]
)
```

```python
cols_to_correlate = continuous_cols + ['Mental_Health_Risk']

# Calculate Spearman correlation
corr_matrix = df[cols_to_correlate].corr(method='spearman')

# Extract correlations with the target
target_corr = corr_matrix[['Mental_Health_Risk']].sort_values(
    by='Mental_Health_Risk'
)

# Plot heatmap
plt.figure(figsize=(8, 7))

sns.heatmap(
    target_corr,
    annot=True,
    cmap='RdBu',
    center=0,
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    cbar_kws={'label': 'Spearman Correlation'}
)

plt.title('Feature Correlation with Mental Health Risk Score',
          fontsize=14, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
```

## EDA Synthesis & Key Findings

### 1. Dataset Overview & Class Balance
* **Dimensions**: 2,000 students and 17 attributes
* **Target Distribution**: `Mental_Health_Risk` is evenly balanced across all 3 classes:
  * **Low**: ~33%
  * **Medium**: ~33%
  * **High**: ~34%
* **Conclusion**: No extreme class imbalance techniques are strictly required

---

### 2. Key Risk & Protective Factors
* **Primary Risk Drivers (+ Correlation / Association)**:
  * **Financial Stress**
  * **Screen Time & Social Media Hours**
  * **Previous Mental Health History**
  * **Diet Quality**
* **Primary Protective Drivers (- Correlation / Association)**:
  * **Sleep Hours**
  * **Family Support**
  * **Physical Activity**

---

### 3. Decisions for Data Preprocessing & Modeling
1. **Identifier Feature**: `Student_ID` carries no predictive information and must be dropped.
2. **Missing Values**:
   * Numerical features: Impute using **Median**
   * Categorical features: Impute using **Mode**
3. **Encoding Strategy**:
   * **Nominal** (`Gender`, `Part_Time_Job`, `Previous_Mental_Health_History`): One-Hot Encoding.
   * **Ordinal** (`Diet_Quality`, `Physical_Activity`): Ordinal Encoding
4. **Feature Scaling**: Numerical features (`Sleep_Hours`, `Financial_Stress`, `GPA`, etc.) will be scaled using `StandardScaler` inside a `ColumnTransformer` to avoid data leakage.
5. **Validation**: Use **Stratified 5-Fold Cross-Validation** to evaluate baseline models (Logistic Regression, Random Forest, Gradient Boosting)
